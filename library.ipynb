```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
data1=pd.read_csv(r"C:\Users\abhir\Downloads\salary.csv")
```

```python
data1.columns
data1.dropna(inplace=True)
data1.drop_duplicates(inplace=True)
data1.fillna(0)
data1.isnull().sum().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
le1=LabelEncoder()
le2=LabelEncoder()
data1['Gender']=le.fit_transform(data1['Gender'])
data1['Education Level']=le1.fit_transform(data1['Education Level'])
data1['Job Title']=le2.fit_transform(data1['Job Title'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x=data1[['Job Title','Education Level','Years of Experience']]
y=data1['Salary']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
model=LinearRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(y_pred)

```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test,y_pred)
```

```python
import pandas as pd
data=pd.read_csv(r"C:\Users\abhir\Downloads\Iris.csv")
print(data)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
x=data[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]]
y=data["Species"]
knn=KNeighborsClassifier(n_neighbors=3)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
knn.fit(x_train,y_train)
y_pred=knn.predict(x_test)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,y_pred))
```

```python
knn.predict([[4.1,3,2.1,0.2]])
```

```python
import pandas as pd
data=pd.read_csv(r"C:\Users\abhir\OneDrive\Desktop\diabetes.csv")
data
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
x=data[["Pregnancies","Glucose","BMI","Age"]]
y = data["Outcome"]
knn = KNeighborsClassifier(n_neighbors=3)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
knn.fit(x_train, y_train)
y_pred = knn.predict(x_test)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test, y_pred))
```

```python
knn.predict([[4.1,3,2.1,0.2]])
```
